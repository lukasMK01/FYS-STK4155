## What you should take away from this project

Beyond the figures and tables in your report, these are the insights we want you to be able to state, explain and defend after working through the project. Use them as a checklist when you write your conclusions.

1.  **A neural network replaces hand-made features by learned ones.** In project 1 the design matrix contained powers of $x$ that you chose. A network takes the raw input and learns its own features in the hidden layers; the universal approximation theorem (Section [8.4](https://educationalmaterialuio.github.io/MachineLearningUiO/doc/LectureNotes/_build/html/chapter8.html#the-universal-approximation-theorem)) tells you that this is possible, not that it is easy.

2.  **Backpropagation is reverse-mode automatic differentiation applied to a layered function.** The four equations of Section [8.8.4](https://educationalmaterialuio.github.io/MachineLearningUiO/doc/LectureNotes/_build/html/chapter8.html#the-four-equations-of-backpropagation) are nothing but the chain rule organized so that one forward and one backward sweep give the gradient with respect to every weight and bias, at a cost of roughly two forward passes, independently of the number of parameters. You should be able to derive them and to check your implementation against `jax.grad`.

3.  **The output activation and the cost function come in matched pairs.** Linear output with the MSE, sigmoid with the binary cross entropy and softmax with the multiclass cross entropy all give the same output error $\boldsymbol{\delta}^L=\boldsymbol{a}^L-\boldsymbol{y}$ (up to the normalization). Mixing the pairs (for instance a sigmoid output with the MSE) gives tiny gradients exactly where the network is confidently wrong.

4.  **The optimizer does not care where the gradient comes from.** Plain gradient descent, stochastic gradient descent, RMSprop and Adam from project 1 carry over unchanged; what changes is that the cost function is no longer convex, so the learning rate, the initialization and the minibatch size now decide whether you converge at all, and to which minimum.

5.  **Initialization and scaling are not details.** Unscaled inputs or weights of the wrong size saturate the sigmoid, and the gradient vanishes (Sections [8.9](https://educationalmaterialuio.github.io/MachineLearningUiO/doc/LectureNotes/_build/html/chapter8.html#vanishing-and-exploding-gradients) and [8.10](https://educationalmaterialuio.github.io/MachineLearningUiO/doc/LectureNotes/_build/html/chapter8.html#weight-initialisation)). Standardized inputs, Xavier or He initialization and the ReLU family exist precisely to keep the gradients alive.

6.  **Hyperparameters must be searched systematically, and on validation data.** The learning rate $\gamma$, the penalty $\lambda$, the depth, the width and the activation function interact. A grid of two parameters at a time, shown as a heat map, is far more informative than a handful of hand-picked runs, and the test set is used only once, at the end.

7.  **Overfitting looks the same as in project 1.** A growing gap between training and test error as the number of parameters grows is the same bias-variance story as for the polynomial degree, and the $L_1$ and $L_2$ penalties play the same role as in Ridge and Lasso regression; early stopping is a form of regularization that costs nothing.

8.  **More flexible is not automatically better.** For a smooth one-dimensional function, OLS with a well-chosen polynomial degree is a closed-form, cheap and hard-to-beat benchmark. The neural network earns its cost only when the features are not known in advance, as for the pixels of MNIST. Your report should say when each method is the right tool.

9.  **A number is meaningless without a baseline.** An accuracy of 97% on MNIST is good only if you know that guessing gives 10%, that logistic regression on raw pixels gives roughly 92% and that convolutional networks reach above 99%. A confusion matrix tells you which errors remain; the accuracy alone does not.

10. **A code you cannot verify is a code you cannot trust.** Gradient checks against automatic differentiation, recovering the closed-form OLS solution with a linear network, and comparing with Scikit-Learn or PyTorch are the tests that turn your implementation into a result you can defend. Seeds and documented hyperparameters make the result reproducible.

```python
# Part a) is analytical and belongs in the theory section of your report.
# You may want to define the cost functions, their gradients and the
# activation functions (with derivatives) here as Python functions.



import numpy as np
import jax
jax.config.update("jax_enable_x64", True)   # 64-bit floats, so that hand and JAX agree to ~1e-15
import jax.numpy as jnp
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(2026)
BLUE, RED, GREEN, GREY = "#004488", "#BB5566", "#228833", "#777777"


def mse(predict, target):
    return jnp.mean((predict - target)**2)


def mse_der(predict, target):
    return 2.0*(predict - target)/predict.size     # the mean brings a factor 2/n



# Defining some activation functions and their derivatives
def ReLU(z):
    return jnp.where(z > 0, z, 0.0)
def ReLU_der(z):
    return jnp.where(z > 0, 1.0, 0.0)

def leaky_ReLU(z, a=0.01):
    return jnp.where(z > 0, z, a * z)

def leaky_ReLU_der(z, a=0.01):
    return jnp.where(z > 0, 1.0, a)

def sigmoid(z):
    return 1 / (1 + jnp.exp(-z))

def sigmoid_der(z):
    s = sigmoid(z)
    return s * (1 - s)

def identity(z):
    return z

def identity_der(z):
    return jnp.ones_like(z)

def softmax(z):
    """Compute softmax values for each set of scores in the rows of the matrix z.
    Used with batched input data: one row of scores per sample."""
    e_z = jnp.exp(z - jnp.max(z, axis=1, keepdims=True))
    return e_z / jnp.sum(e_z, axis=1, keepdims=True)
```

```python
def MSE(y_data, y_model):
    """from Tuesday lecture week 35"""
    return np.mean((y_data - y_model) ** 2)

def R2(y_data, y_model):
    """from Tuesday lecture week 35"""
    return 1.0 - np.sum((y_data - y_model) ** 2) / np.sum((y_data - np.mean(y_data)) ** 2)

# Functions from exercises
def cost(theta, X, y, lam=0.0):
    #if lambda == 0, this is OLS, otherwise Ridge
    return jnp.mean((y - X @ theta)**2) + lam * jnp.sum(theta**2)

def closed_form(X, y, lam=0.0):
    """Eq. (3.44) with the 1/n convention of Eq. (3.95): (X^T X + n lambda I)^-1 X^T y."""
    n, p = X.shape
    return np.linalg.solve(X.T @ X + n * lam * np.eye(p), X.T @ y)

def gradient(theta, X, y, lam=0.0):
    """Eqs. (4.13) and (4.17): the gradient of (1/n)||X theta - y||^2 + lambda theta^T theta."""
    n = len(y)
    return (2.0 / n) * X.T @ (X @ theta - y) + 2.0 * lam * theta
```

```python
# Your code for part b) here

#Functions to do the backpropagation and feed forward

def create_layers(network_input_size, layer_output_sizes):
    layers = []

    i_size = network_input_size
    for layer_output_size in layer_output_sizes:
        W = np.random.randn(layer_output_size, i_size)
        b = np.random.randn(layer_output_size)
        layers.append((W, b))

        i_size = layer_output_size
    return layers

def feed_forward(input, layers, activation_funcs):
    a = input
    for (W, b), activation_func in zip(layers, activation_funcs):
        z = W @ a + b
        a = activation_func(z)
    return a

def cost(layers, input, activation_funcs, target):
    predict = feed_forward(input, layers, activation_funcs)
    return mse(predict, target)

def feed_forward_saver(input, layers, activation_funcs):
    layer_inputs = []
    zs = []
    a = input
    for (W, b), activation_func in zip(layers, activation_funcs):
        layer_inputs.append(a)
        z = W @ a + b
        a = activation_func(z)

        zs.append(z)

    return layer_inputs, zs, a



def backpropagation(
    input, layers, activation_funcs, target, activation_ders, cost_der=mse_der
):
    layer_inputs, zs, predict = feed_forward_saver(input, layers, activation_funcs)

    layer_grads = [() for layer in layers]

    # We loop over the layers, from the last to the first
    for i in reversed(range(len(layers))):
        layer_input, z, activation_der = layer_inputs[i], zs[i], activation_ders[i]

        if i == len(layers) - 1:
            # For last layer we use cost derivative as dC_da(L) can be computed directly
            dC_da = cost_der(predict, target)
        else:
            # For other layers we build on previous z derivative, as dC_da(i) = dC_dz(i+1) * dz(i+1)_da(i)
            (W, b) = layers[i + 1]
            dC_da = W.T @ dC_dz

        dC_dz = dC_da * activation_der(z)
        dC_dW = np.outer(dC_dz, layer_input)
        dC_db = dC_dz

        layer_grads[i] = (dC_dW, dC_db)

    return layer_grads

def create_layers_batch(network_input_size, layer_output_sizes):
    layers = []
    i_size = network_input_size
    for layer_output_size in layer_output_sizes:
        W = np.random.randn(i_size, layer_output_size)
        b = np.random.randn(layer_output_size)
        layers.append((W, b))
        i_size = layer_output_size
    return layers


def feed_forward_batch(inputs, layers, activation_funcs):
    a = inputs
    for (W, b), activation_func in zip(layers, activation_funcs):
        a = activation_func(a @ W + b)
    return a


def feed_forward_saver_batch(inputs, layers, activation_funcs):
    layer_inputs, zs = [], []
    a = inputs
    for (W, b), activation_func in zip(layers, activation_funcs):
        layer_inputs.append(a)
        z = a @ W + b
        a = activation_func(z)
        zs.append(z)
    return layer_inputs, zs, a


def mse_der_batch(predict, target):
    return 2 * (predict - target) / predict.size


def cost_batch(layers, inputs, activation_funcs, targets):
    return mse(feed_forward_batch(inputs, layers, activation_funcs), targets)

def output_error_softmax_ce(predict, target):
    return predict - target          # delta^L = a^L - y, Eq. (8.69)


def backpropagation_batch(
    inputs, layers, activation_funcs, targets, activation_ders, cost_der=mse_der_batch, output_error=None,
):
    layer_inputs, zs, predict = feed_forward_saver_batch(inputs, layers, activation_funcs)
    layer_grads = [() for layer in layers]

    for i in reversed(range(len(layers))):
        layer_input, z, activation_der = layer_inputs[i], zs[i], activation_ders[i]
        if i == len(layers) - 1 and output_error is not None:
            dC_dz = output_error(predict, targets)          # the shortcut: softmax + cross entropy
        else:
            if i == len(layers) - 1:
                dC_da = cost_der(predict, targets)
            else:
                (W, b) = layers[i + 1]
                dC_da = dC_dz @ W.T                 # samples stay in the rows
            dC_dz = dC_da * activation_der(z)       # (n, n_out), one error vector per sample
        dC_dW = layer_input.T @ dC_dz           # sum over the samples of the outer products
        dC_db = np.sum(dC_dz, axis=0)           # sum over the samples of the errors
        layer_grads[i] = (dC_dW, dC_db)

    return layer_grads
```

```python
# standardised features and an 80/20 split, as in the week-41 solutions
from sklearn.model_selection import train_test_split

#initialise the weights and biases
def init_layers(sizes, seed):
    rng = np.random.default_rng(seed)
    return [(rng.standard_normal((n_in, n_out)) / np.sqrt(n_in), np.zeros(n_out)) #weights are initialised with a normal distribution with mean 0 and variance 1/n_in, biases are initialised to 0
            for n_in, n_out in zip(sizes[:-1], sizes[1:])]

#using MSE as the cost function:
def cost_batch(layers, inputs, activation_funcs, targets):
    return mse(feed_forward_batch(inputs, layers, activation_funcs), targets)


#setting up the input and target data:
def prepare_data(x,y):
    """splits into train and test, prepares design matrix and scales data"""

    #split into train and test data
    x_tr, x_te, y_tr, y_te = train_test_split(x, y, test_size=0.2, random_state=2026)

    # create design matrix, expanding the dimensions of x and y to make them 2D arrays with one column each
    X_train = np.expand_dims(x_tr, axis=1)
    X_test = np.expand_dims(x_te, axis=1) 

    #need to expand the dimensions of y as well, to make it a 2D array with one column
    y_tr = np.expand_dims(y_tr, axis=1)
    y_te = np.expand_dims(y_te, axis=1)

    #centering and scaling the data
    x_tr_std = (X_train - X_train.mean(axis=0)) / X_train.std(axis=0)
    y_tr_norm = y_tr# - y_tr.mean(axis=0)

    x_te_std = (X_test - X_train.mean(axis=0)) / X_train.std(axis=0)
    y_te_norm = y_te# - y_te.mean(axis=0)

    return x_tr_std, x_te_std, y_tr_norm, y_te_norm


#From weekly exercises, incorporated into the optimiser_step function
def gd_step(layers, grads, gamma, state=None):
    """Plain gradient descent on a list of (W, b): theta <- theta - gamma * grad."""
    return [(W - gamma * dW, b - gamma * db) for (W, b), (dW, db) in zip(layers, grads)], state

def momentum_step(layers, grads, gamma, state=None, beta=0.9):
    """SGD with momentum: v <- beta v + gamma grad, theta <- theta - v.
    The velocities live in `state`, one (vW, vb) per layer."""
    if state is None:
        state = [(np.zeros_like(W), np.zeros_like(b)) for W, b in layers]
    state = [(beta * vW + gamma * dW, beta * vb + gamma * db) for (vW, vb), (dW, db) in zip(state, grads)]
    layers = [(W - vW, b - vb) for (W, b), (vW, vb) in zip(layers, state)]
    return layers, state


#Function from exercises, equations for various methods with gradient descent 
#help from copilot with in-line suggestions based on the original function with linear regression, that was looked through before (if) accepted
def optimiser_step(method, layers, grads, gamma, t,state=None, beta=0.9, rho=0.99,
                   beta1=0.9, beta2=0.999, eps=1e-8):
    """One update of theta from the gradient g at step t (t = 1, 2, ...), Eqs. (4.10), (4.28),
    (4.42)-(4.45), (4.47)-(4.48) and (4.51)-(4.55).  state carries the running quantities."""
    if method == "plain":
        return [(W - gamma * dW, b - gamma * db) for (W, b), (dW, db) in zip(layers, grads)], state #theta - gamma * g, state
    if method == "momentum":
        if state is None:
            state = [(np.zeros_like(W), np.zeros_like(b)) for W, b in layers]
        state = [(beta * vW + gamma * dW, beta * vb + gamma * db) for (vW, vb), (dW, db) in zip(state, grads)]
        layers = [(W - vW, b - vb) for (W, b), (vW, vb) in zip(layers, state)]
        return layers, state
    if method == "rmsprop":
        if state is None:
            state = [(np.zeros_like(W), np.zeros_like(b)) for W, b in layers]
        state = [(rho * rW + (1.0 - rho) * dW * dW, rho * rb + (1.0 - rho) * db * db) for (rW, rb), (dW, db) in zip(state, grads)]
        layers = [(W - gamma * dW / (np.sqrt(rW) + eps), b - gamma * db / (np.sqrt(rb) + eps)) for (W, b), (dW, db), (rW, rb) in zip(layers, grads, state)]
        return layers, state  
    if method == "adam": #help from local chat to debug the code and come with suggestions for fix for adam implementation
        if state is None:
            m_state = [(np.zeros_like(W), np.zeros_like(b)) for W, b in layers]
            r_state = [(np.zeros_like(W), np.zeros_like(b)) for W, b in layers]      
        else:
            m_state, r_state = state
        m_state = [(beta1 * mW + (1.0 - beta1) * dW, beta1 * mb + (1.0 - beta1) * db) for (mW, mb), (dW, db) in zip(m_state, grads)]
        r_state = [(beta2 * rW + (1.0 - beta2) * dW * dW, beta2 * rb + (1.0 - beta2) * db * db) for (rW, rb), (dW, db) in zip(r_state, grads)]
        state = [m_state,r_state]
        layers = [(W - gamma * (mW / (1.0 - beta1**t)) / (np.sqrt(rW / (1.0 - beta2**t)) + eps), b - gamma * (mb / (1.0 - beta1**t)) / (np.sqrt(rb / (1.0 - beta2**t)) + eps)) for (W, b), (mW, mb), (rW, rb) in zip(layers, m_state, r_state)]
        return layers, state 
    raise ValueError(f"unknown method {method}")


def OLS_regression(X, y):
    """ set up the design function, based on a 5th degree polynomial, and return the closed form solution with y tilde as the predicted target"""
    theta = closed_form(np.column_stack([X**i for i in range(6)]), y)

    predict_OLS = np.polyval(theta[::-1], X)

    return predict_OLS

def predict(X, layers, hidden_activation="sigmoid"):
    """The output of the network"""
    return feed_forward_batch(X, layers, hidden_activation)



#function from exercises, tweaked to match the project
def train(layers, X, y, X_test, y_test, step, gamma, epochs, batch_size=None, seed=2026):
    """batch_size=None: full-batch gradient descent; otherwise minibatches drawn without
    replacement from a fresh shuffle every epoch.  Returns the trained layers and a history
    of (epoch, training cost, test accuracy)."""
    rng = np.random.default_rng(seed)
    n = len(X)
    M = n if batch_size is None else batch_size
    state, history = None, []
    t = 0
    ytilde = []
    for epoch in range(1, epochs + 1):
        perm = rng.permutation(n)
        for start in range(0, n, M):
            
            idx = perm[start:start + M]
            grads = backpropagation_batch(X[idx], layers, activation_funcs, y[idx], activation_ders)

            t += 1
            layers, state = optimiser_step(step, layers, grads, gamma, t, state)

        #Getting the target values from closed form solution of OLS to compare with the neural network
        ytilde_OLS = OLS_regression(X_test, y_test)

        #The MSE as cost, and the convergence to closed form OLS
        history.append((epoch, float(cost_batch(layers, X, activation_funcs, y)),float(np.linalg.norm(feed_forward_batch(X_test, layers, activation_funcs) - ytilde_OLS))))
        ytilde.append(feed_forward_batch(X_test, layers, activation_funcs)) 
    return layers, np.array(history), ytilde


def setup_NN(
    X_shape,
    y_shape,
    design_hidden:list,
    activation_funcs:callable=ReLU,
    activation_ders:callable=ReLU_der,
    activation_output: callable=identity,
    activation_output_der: callable=identity_der,
):
    """Set up a neural network with the layers and activation functions."""
    layers = init_layers([X_shape[1]] + design_hidden + [y_shape[1]], seed=2026)
    activation_funcs = [activation_funcs] * len(design_hidden) + [activation_output]
    activation_ders = [activation_ders] * len(design_hidden) + [activation_output_der]

    return layers, activation_funcs, activation_ders

```
